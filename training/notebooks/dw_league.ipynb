# DEADWEIGHT league training
One cell: clones the repo, builds `dw_server`, installs the RL deps, resumes each role from the IDUNA registry (if you give it the DEADWEIGHT-RL secret) and trains the 3-role PFSP league, pushing every generation back. Blank secret = local-only run.
Same shape as BRAWLPIT's `scripts/colab_train.py`. Tunables: `DEADWEIGHT_GENERATIONS`, `DEADWEIGHT_TIMESTEPS`, `DEADWEIGHT_SAVE_FREQ`.
